# Match Possession Analysis

# Step 1: Create the function

```python
# ============================================================
# MATCH POSSESSION ANALYSIS
# ============================================================

from collections import Counter

def calculate_match_possession(frames):

    possession_counts = Counter()

    # ---------------------------------------
    # Count possession frames
    # ---------------------------------------

    for frame in frames:

        possession = frame.get("possession", {})
        team = possession.get("group")

        if team == "home team":
            possession_counts["home"] += 1

        elif team == "away team":
            possession_counts["away"] += 1

        else:
            possession_counts["none"] += 1

    # ---------------------------------------
    # Frame totals
    # ---------------------------------------

    total_frames = (
        possession_counts["home"] +
        possession_counts["away"] +
        possession_counts["none"]
    )

    possession_frames = (
        possession_counts["home"] +
        possession_counts["away"]
    )

    # ---------------------------------------
    # Possession percentages
    # ---------------------------------------

    if possession_frames > 0:

        home_percent = (
            100 * possession_counts["home"] /
            possession_frames
        )

        away_percent = (
            100 * possession_counts["away"] /
            possession_frames
        )

    else:

        home_percent = 0
        away_percent = 0

    # ---------------------------------------
    # Tracking diagnostic
    # ---------------------------------------

    no_possession_percent = (
        100 * possession_counts["none"] /
        total_frames
    )

    # ---------------------------------------
    # Return results
    # ---------------------------------------

    return {

        "home_frames": possession_counts["home"],
        "away_frames": possession_counts["away"],
        "no_possession_frames": possession_counts["none"],

        "home_percent": home_percent,
        "away_percent": away_percent,

        "no_possession_percent": no_possession_percent,

        "total_frames": total_frames,
        "possession_frames": possession_frames
    }
```

```python
possession = calculate_match_possession(frames)

print("=" * 60)
print("MATCH POSSESSION ANALYSIS")
print("=" * 60)

print("\nOfficial Possession")
print(f"{home_team:<30} {possession['home_percent']:.2f}%")
print(f"{away_team:<30} {possession['away_percent']:.2f}%")

print("\nTracking Diagnostics")
print(f"Home Frames              : {possession['home_frames']:,}")
print(f"Away Frames              : {possession['away_frames']:,}")
print(f"No Possession Frames     : {possession['no_possession_frames']:,}")
print(f"Total Frames             : {possession['total_frames']:,}")
print(f"No Possession Rate       : {possession['no_possession_percent']:.2f}%")
```

# Distance Covered


# Create the function

```python
# ============================================================
# MATCH DISTANCE COVERED
# ============================================================

import math
from collections import defaultdict

def calculate_match_distance(frames):

    # Total distance covered by each player (metres)
    distance = defaultdict(float)

    # Previous position of each player
    previous_position = {}

    for frame in frames:

        for player in frame["player_data"]:

            # Ignore extrapolated positions
            if not player["is_detected"]:
                continue

            player_id = player["player_id"]

            x = player["x"]
            y = player["y"]

            # First observation
            if player_id not in previous_position:

                previous_position[player_id] = (x, y)
                continue

            prev_x, prev_y = previous_position[player_id]

            # Euclidean distance
            d = math.sqrt(
                (x - prev_x) ** 2 +
                (y - prev_y) ** 2
            )

            distance[player_id] += d

            previous_position[player_id] = (x, y)

    return dict(distance)
```

# Sprint Analysis

# Create the function

```python
# ============================================================
# MATCH SPRINT ANALYSIS
# ============================================================

import math
from collections import defaultdict

FPS = 10
SPRINT_THRESHOLD = 25.0      # km/h
MAX_VALID_SPEED = 38.5       # km/h
MIN_MOVEMENT = 0.05          # metres


def calculate_sprints(frames):

    sprint_count = defaultdict(int)
    max_speed = defaultdict(float)

    # Store previous x, y and frame number
    previous_position = {}

    # Track whether a player is currently sprinting
    is_sprinting = defaultdict(bool)

    for frame in frames:

        current_frame = frame["frame"]

        for player in frame["player_data"]:

            # Ignore extrapolated positions
            if not player["is_detected"]:
                continue

            player_id = player["player_id"]

            x = player["x"]
            y = player["y"]

            # First observation of this player
            if player_id not in previous_position:

                previous_position[player_id] = (
                    x,
                    y,
                    current_frame
                )

                continue

            # Previous position
            prev_x, prev_y, prev_frame = previous_position[player_id]

            # ----------------------------------------------------
            # Distance travelled (metres)
            # ----------------------------------------------------

            distance = math.sqrt(
                (x - prev_x) ** 2 +
                (y - prev_y) ** 2
            )

            # Ignore tiny tracking jitter
            if distance < MIN_MOVEMENT:
                distance = 0

            # ----------------------------------------------------
            # Time elapsed
            # ----------------------------------------------------

            frame_gap = current_frame - prev_frame

            if frame_gap <= 0:

                previous_position[player_id] = (
                    x,
                    y,
                    current_frame
                )

                continue

            time_elapsed = frame_gap / FPS

            # ----------------------------------------------------
            # Speed (km/h)
            # ----------------------------------------------------

            speed = (distance / time_elapsed) * 3.6

            # Ignore impossible tracking jumps
            if speed > MAX_VALID_SPEED:

                previous_position[player_id] = (
                    x,
                    y,
                    current_frame
                )

                continue

            # Record maximum speed
            if speed > max_speed[player_id]:
                max_speed[player_id] = speed

            # ----------------------------------------------------
            # Sprint Detection
            # ----------------------------------------------------

            if speed >= SPRINT_THRESHOLD:

                # Count only when a sprint starts
                if not is_sprinting[player_id]:

                    sprint_count[player_id] += 1
                    is_sprinting[player_id] = True

            else:

                # Sprint has ended
                is_sprinting[player_id] = False

            # Update previous position
            previous_position[player_id] = (
                x,
                y,
                current_frame
            )

    return sprint_count, max_speed
```

# Team Width Analysis
Team Width = Right-most player − Left-most player

# Create the Team Width Function

```python
# ============================================================
# TEAM WIDTH ANALYSIS
# ============================================================

import numpy as np

def calculate_team_width(frames, team_id):

    widths = []

    for frame in frames:

        y_positions = []

        for player in frame["player_data"]:

            # Ignore undetected players
            if not player["is_detected"]:
                continue

            info = player_lookup.get(player["player_id"])

            if info is None:
                continue

            if info["team_id"] != team_id:
                continue

            # Use x-coordinate (pitch length direction)
            y_positions.append(player["y"])

        # Need enough players to calculate width
        if len(y_positions) < 4:
            continue

        width = max(y_positions) - min(y_positions)

        widths.append(width)

    return widths
```

```python
# ============================================================
# TEAM LENGTH ANALYSIS
# ============================================================

import numpy as np

MIN_PLAYERS = 9      # Minimum detected players required

def calculate_team_length(frames, team_id):

    lengths = []

    for frame in frames:

        x_positions = []

        for player in frame["player_data"]:

            # Ignore undetected players
            if not player["is_detected"]:
                continue

            info = player_lookup.get(player["player_id"])

            if info is None:
                continue

            # Keep only this team's players
            if info["team_id"] != team_id:
                continue

            x_positions.append(player["x"])

        # Ignore poor tracking frames
        if len(x_positions) < MIN_PLAYERS:
            continue

        team_length = max(x_positions) - min(x_positions)

        lengths.append(team_length)

    return lengths
```

How much space does the team occupy?

# Calculate Convex Hull Area

```python
# ============================================================
# CONVEX HULL AREA
# ============================================================

from scipy.spatial import ConvexHull

MIN_PLAYERS = 9

def calculate_convex_hull_area(frames, team_id):

    areas = []

    for frame in frames:

        points = []

        for player in frame["player_data"]:

            # Ignore undetected players
            if not player["is_detected"]:
                continue

            info = player_lookup.get(player["player_id"])

            if info is None:
                continue

            if info["team_id"] != team_id:
                continue

            points.append([
                player["x"],
                player["y"]
            ])

        # Require enough detected players
        if len(points) < MIN_PLAYERS:
            continue

        points = np.array(points)

        hull = ConvexHull(points)

        areas.append(hull.volume)

    return areas
```

# Team Compactness
Team compactness measures how closely connected a team's players are.

# Calculate Compactness

```python
# ============================================================
# TEAM SHAPE ANALYSIS
# Width + Length + Convex Hull + Compactness
# ============================================================

import numpy as np
from scipy.spatial import ConvexHull

MIN_PLAYERS = 9

def calculate_team_shape(frames, team_id):

    results = []

    for frame in frames:

        points = []

        for player in frame["player_data"]:

            if not player["is_detected"]:
                continue

            info = player_lookup.get(player["player_id"])

            if info is None:
                continue

            if info["team_id"] != team_id:
                continue

            points.append((player["x"], player["y"]))

        # Ignore poor tracking frames
        if len(points) < MIN_PLAYERS:
            continue

        points = np.array(points)

        x = points[:, 0]
        y = points[:, 1]

        # -------------------------
        # Width & Length
        # -------------------------

        width = np.max(y) - np.min(y)
        length = np.max(x) - np.min(x)

        # -------------------------
        # Convex Hull
        # -------------------------

        hull = ConvexHull(points)
        hull_area = hull.volume

        # -------------------------
        # Compactness
        # -------------------------

        rectangle_area = width * length

        if rectangle_area == 0:
            continue

        compactness = hull_area / rectangle_area

        results.append({

            "frame": frame["frame"],
            "timestamp": frame["timestamp"],

            "width": width,
            "length": length,

            "hull_area": hull_area,
            "compactness": compactness

        })

    return results
```

# Calculate Both Teams

```python
home_shape = calculate_team_shape(
    frames,
    home_team_id
)

away_shape = calculate_team_shape(
    frames,
    away_team_id
)
```

# Extract the metrics

```python
home_widths = [x["width"] for x in home_shape]
home_lengths = [x["length"] for x in home_shape]
home_hull = [x["hull_area"] for x in home_shape]
home_compactness = [x["compactness"] for x in home_shape]

away_widths = [x["width"] for x in away_shape]
away_lengths = [x["length"] for x in away_shape]
away_hull = [x["hull_area"] for x in away_shape]
away_compactness = [x["compactness"] for x in away_shape]
```

# Space Control
How much of the pitch each team controls at any moment?




# Create the Pitch Grid

```python
import numpy as np

GRID_SIZE = 2

x_grid = np.arange(-52.5, 52.5, GRID_SIZE)
y_grid = np.arange(-34, 34, GRID_SIZE)
```

# Space Control Function

```python
# ============================================================
# SPACE CONTROL
# ============================================================

import numpy as np

MIN_PLAYERS = 9

def calculate_space_control(frames):

    results = []

    for frame in frames:

        home_players = []
        away_players = []

        for player in frame["player_data"]:

            if not player["is_detected"]:
                continue

            info = player_lookup.get(player["player_id"])

            if info is None:
                continue

            point = np.array([player["x"], player["y"]])

            if info["team_id"] == home_team_id:
                home_players.append(point)
            else:
                away_players.append(point)

        if len(home_players) < MIN_PLAYERS:
            continue

        if len(away_players) < MIN_PLAYERS:
            continue

        home_cells = 0
        away_cells = 0

        for x in x_grid:
            for y in y_grid:

                location = np.array([x, y])

                home_distance = min(
                    np.linalg.norm(location - p)
                    for p in home_players
                )

                away_distance = min(
                    np.linalg.norm(location - p)
                    for p in away_players
                )

                if home_distance < away_distance:
                    home_cells += 1
                else:
                    away_cells += 1

        total = home_cells + away_cells

        results.append({

            "frame": frame["frame"],
            "timestamp": frame["timestamp"],

            "home_space": 100 * home_cells / total,
            "away_space": 100 * away_cells / total

        })

    return results
```
